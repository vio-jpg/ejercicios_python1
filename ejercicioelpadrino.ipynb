```python
# =====================================================================
# BASE DE DATOS DE PELÍCULAS (Vectores de Género)
# Orden del vector: [Acción, Comedia, Ciencia Ficción, Drama, Suspenso]
# =====================================================================

peliculas = {
    "Matrix": [1, 0, 1, 0, 1],
    "El Padrino": [0, 0, 0, 1, 1],
    "Interstellar": [0, 0, 1, 1, 0],
    "Toy Story": [0, 1, 0, 0, 0],
    "Avengers": [1, 1, 1, 0, 0],
    "Superbad": [0, 1, 0, 0, 0],
    "Inception": [1, 0, 1, 0, 1],
    "Gladiador": [1, 0, 0, 1, 0],
    "Parasite": [0, 1, 0, 1, 1],
    "Batman: El Caballero de la Noche": [1, 0, 0, 1, 1],
    "Buscando a Nemo": [0, 1, 0, 0, 0]
}

# Diccionario auxiliar para buscar por palabras clave (como un filtro de spam)
palabras_clave_peliculas = {
    "Matrix": ["accion", "ciencia ficcion", "simulacion", "computadoras", "suspenso"],
    "El Padrino": ["drama", "mafia", "crimen", "clasico", "suspenso"],
    "Interstellar": ["ciencia ficcion", "espacio", "drama", "tiempo", "estrellas"],
    "Toy Story": ["comedia", "animacion", "juguetes", "ninos", "amigos"],
    "Avengers": ["accion", "superheroes", "marvel", "ciencia ficcion", "comedia"],
    "Superbad": ["comedia", "amigos", "fiesta", "adolescentes", "risa"],
    "Inception": ["ciencia ficcion", "suenos", "accion", "suspenso", "mente"],
    "Gladiador": ["accion", "drama", "roma", "historia", "venganza"],
    "Parasite": ["drama", "suspenso", "comedia negra", "corea", "intriga"],
    "Batman: El Caballero de la Noche": ["accion", "superheroes", "drama", "suspenso", "joker"],
    "Buscando a Nemo": ["comedia", "animacion", "pez", "oceano", "familia"]
}

# =====================================================================
# FUNCIONES DEL SCRIPT
# =====================================================================

def calcular_similitud(peli1, peli2):
    """Calcula el producto punto entre los vectores de dos películas."""
    vec1 = peliculas[peli1]
    vec2 = peliculas[peli2]
    return sum([a * b for a, b in zip(vec1, vec2)])


def buscar_por_palabra_clave(palabra):
    """Busca películas que contengan la palabra clave (filtro tipo spam)."""
    palabra = palabra.lower().strip()
    resultados = []

    for peli, keywords in palabras_clave_peliculas.items():
        # Si la palabra clave está en la lista de keywords o en el propio título
        if palabra in keywords or palabra in peli.lower():
            resultados.append(peli)

    return resultados


def mostrar_peliculas():
    """Muestra la lista de películas disponibles."""
    print("\n--- Películas Disponibles ---")
    for i, peli in enumerate(peliculas.keys(), 1):
        print(f"{i}. {peli}")
    print("-----------------------------\n")

# =====================================================================
# FLUJO PRINCIPAL DEL PROGRAMA (Interfaz de usuario)
# =====================================================================

def menu():
    while True:
        print("=== SISTEMA DE RECOMENDACIÓN E IA DE PELÍCULAS ===")
        print("1. Calcular similitud entre dos películas")
        print("2. Buscar películas por palabra clave (Filtro tipo Spam)")
        print("3. Salir")

        opcion = input("Elige una opción (1-3): ")

        if opcion == "1":
            mostrar_peliculas()
            peli1 = input("Escribe el nombre exacto de la primera película: ")
            peli2 = input("Escribe el nombre exacto de la segunda película: ")

            if peli1 in peliculas and peli2 in peliculas:
                similitud = calcular_similitud(peli1, peli2)
                print(f"\n[RESULTADO] La similitud (coincidencia de géneros) entre '{peli1}' y '{peli2}' es: {similitud}\n")
            else:
                print("\n[ERROR] Una o ambas películas no existen en la base de datos. Inténtalo de nuevo.\n")

        elif opcion == "2":
            palabra = input("\nIntroduce una palabra clave (ej. 'espacio', 'accion', 'animacion'): ")
            resultados = buscar_por_palabra_clave(palabra)

            if resultados:
                print(f"\n[RESULTADOS] Películas encontradas para '{palabra}':")
                for peli in resultados:
                    print(f" - {peli}")
                print()
            else:
                print(f"\n[INFO] No se encontraron películas con la palabra '{palabra}'.\n")

        elif opcion == "3":
            print("\n¡Gracias por usar el script de IA! Hasta luego.")
            break
        else:
            print("\nOpción no válida. Por favor, elige 1, 2 o 3.\n")

# Ejecutar el programa
if __name__ == "__main__":
    menu()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Carga del dataset
df = pd.read_csv('notas_estudiantes_completo.csv')

# 1. Barplot: Nota media por asignatura
sns.barplot(x='asignatura', y='nota', data=df)
plt.title('Nota Media por Asignatura')
plt.ylim(0,10)
plt.show()

# 2. Boxplot: Dispersión de las notas
sns.boxplot(x='asignatura', y='nota', data=df)
plt.title('Dispersión y Variabilidad de Notas')
plt.ylim(1,10)
plt.show()

# 3. Countplot: Cantidad de registros por asignatura
sns.countplot(x='asignatura', data=df)
plt.title('Número de Registros por Asignatura')
plt.show()
```

```python
# 4. Heatmap: Notas por Estudiante y Asignatura
plt.figure(figsize=(10, 8))

# Es necesario pivotar el DataFrame para tener asignaturas en columnas y estudiantes en filas
df_pivot = df.pivot(index='estudiante', columns='asignatura', values='nota')

sns.heatmap(df_pivot, annot=True, cmap='RdYlGn', linewidths=.5)
plt.title('Mapa de Calor: Notas por Estudiante')
plt.xlabel('Asignatura')
plt.ylabel('Estudiante')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 1. Cargar el dataset
df = pd.read_csv('notas_estudiantes.csv')

# Asegurar que la columna 'fecha' es datetime
df['fecha'] = pd.to_datetime(df['fecha'])

# --- PARTE 1: Gráfico de promedios generales ---

# 2. Calcular promedio de notas por asignatura
promedios = df.groupby('asignatura')['nota'].mean()

# 3. Gráfico de barras: promedio por asignatura
plt.figure(figsize=(10, 5))
plt.bar(promedios.index, promedios.values, color='skyblue')
plt.title("Promedio de notas por asignatura")
plt.xlabel("Asignatura")
plt.ylabel("Nota media")
plt.ylim(0, 10)

for i, valor in enumerate(promedios.values):
    plt.text(i, valor + 0.1, f"{valor:.1f}", ha='center')

plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.show()

# --- PARTE 2: Gráfico de un estudiante específico ---

# Elegir un estudiante específico
estudiante = 'Estudiante_5'
df_est = df[df['estudiante'] == estudiante]

# Crear gráfico de líneas
plt.figure(figsize=(10, 5))
plt.plot(df_est['asignatura'], df_est['nota'], marker='o', linestyle='-', color='green')
plt.title(f"Notas de {estudiante}")
plt.xlabel("Asignatura")
plt.ylabel("Nota")
plt.ylim(0, 10)

for i, nota in enumerate(df_est['nota']):
    plt.text(i, nota + 0.1, f"{nota:.1f}", ha='center')

plt.grid(True)
plt.show()
```
